```python
from bs4 import BeautifulSoup
import requests

url = 'https://firms.modaps.eosdis.nasa.gov/active_fire/' 
with open('html.txt','r') as f:
    html = f.read()
print(html)
soup = BeautifulSoup(html, 'html.parser')

seven_day_links = []
print(soup.prettify())
for link in soup.find_all('a', href=True):
    if '7d' in link['href'] and link['href'].endswith('.csv'):
        seven_day_links.append(link['href'])

for link in seven_day_links:
    print(link)
```
