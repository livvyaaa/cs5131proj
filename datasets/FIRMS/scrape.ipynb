```python
from bs4 import BeautifulSoup
import requests
```

```python

with open('html.txt','r') as f:
    html = f.read()
soup = BeautifulSoup(html, 'html.parser')

seven_day_links = []
for link in soup.find_all('a', href=True):
    if '7d' in link['href'] and link['href'].endswith('.csv'):
        seven_day_links.append(link['href'][1:])

print(seven_day_links)
```

```python
base_url = 'https://firms.modaps.eosdis.nasa.gov/' 
for link in seven_day_links:
    url = base_url + link
    print(url)
    r = requests.get(url)
    filename = link.split('/')[-1]
    with open(filename, 'wb') as f:
        f.write(r.content)
    print('Downloaded', link)
```
